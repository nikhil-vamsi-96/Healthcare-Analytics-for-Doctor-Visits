# 🏥 Healthcare Analytics for Doctor Visits

doctor visits, illness, health conditions, reduced activity,

The objective is to explore the dataset, identify important patterns,
and communicate meaningful healthcare insights through Python

- Perform data cleaning and preprocessing
- Analyze patient-related variables
- Analyze doctor visit patterns
- Study illness and health conditions
- Analyze reduced activity
- Analyze healthcare coverage
- Analyze chronic conditions
- Study relationships between numerical variables
- Create meaningful visualizations
- Summarize key findings

## Technologies Used

- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn
- Jupyter Notebook / Google Colab

```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Visualization style
sns.set_style("whitegrid")

# Display options
pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

```python
# Upload dataset in Google Colab

from google.colab import files

uploaded = files.upload()

print("Dataset uploaded successfully!")
```

```python
# Load uploaded CSV file

import io

filename = list(uploaded.keys())[0]

df = pd.read_csv(io.BytesIO(uploaded[filename]))

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
# Remove unnecessary index column

if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

print("Dataset after cleaning:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
# Display dataset information

print("Dataset Information")
print("=" * 50)

df.info()
```

```python
# Number of rows and columns

print("Number of records:", df.shape[0])
print("Number of variables:", df.shape[1])
```

```python
# Display column names

print("Columns in the dataset:")
print("=" * 50)

for number, column in enumerate(df.columns, start=1):
    print(f"{number}. {column}")
```

```python
# Display first 10 records

display(df.head(10))
```

```python
# Display last 10 records

display(df.tail(10))
```

```python
# Descriptive statistics for numerical variables

display(df.describe())
```

```python
# Check missing values

missing_values = df.isnull().sum()

missing_summary = pd.DataFrame({
    "Column": missing_values.index,
    "Missing Values": missing_values.values
})

display(missing_summary)
```

```python
# Check duplicate records

duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```

```python
# Display data types

data_types = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values
})

display(data_types)
```

```python
# Examine unique values of categorical variables

categorical_columns = [
    'gender',
    'private',
    'freepoor',
    'freerepat',
    'nchronic',
    'lchronic'
]

for column in categorical_columns:
    print(f"\n{column.upper()}")
    print(df[column].value_counts())
```

```python
# Gender distribution

gender_counts = df['gender'].value_counts()

display(gender_counts)

plt.figure(figsize=(7, 5))

sns.countplot(
    x='gender',
    data=df
)

plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Number of Records')

plt.show()
```

```python
# Distribution of doctor visits

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x='visits',
    bins=10,
    kde=True
)

plt.title('Distribution of Doctor Visits')
plt.xlabel('Number of Doctor Visits')
plt.ylabel('Frequency')

plt.show()
```

```python
# Frequency of doctor visits

visit_frequency = (
    df['visits']
    .value_counts()
    .sort_index()
    .reset_index()
)

visit_frequency.columns = [
    'Number of Visits',
    'Number of Patients'
]

display(visit_frequency)
```

```python
# Distribution of age variable

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x='age',
    bins=20,
    kde=True
)

plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.show()
```

```python
# Distribution of income

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x='income',
    bins=20,
    kde=True
)

plt.title('Income Distribution')
plt.xlabel('Income')
plt.ylabel('Frequency')

plt.show()
```

```python
# Distribution of illness levels

illness_counts = df['illness'].value_counts().sort_index()

display(illness_counts)

plt.figure(figsize=(8, 5))

sns.countplot(
    x='illness',
    data=df
)

plt.title('Distribution of Number of Illnesses')
plt.xlabel('Number of Illnesses')
plt.ylabel('Number of Records')

plt.show()
```

```python
# Distribution of reduced activity

plt.figure(figsize=(9, 5))

sns.countplot(
    x='reduced',
    data=df
)

plt.title('Distribution of Reduced Activity')
plt.xlabel('Reduced Activity')
plt.ylabel('Number of Records')

plt.show()
```

```python
# Distribution of health condition

plt.figure(figsize=(9, 5))

sns.countplot(
    x='health',
    data=df
)

plt.title('Distribution of Health Condition')
plt.xlabel('Health Condition')
plt.ylabel('Number of Records')

plt.show()
```

```python
# Healthcare coverage variables

coverage_columns = [
    'private',
    'freepoor',
    'freerepat'
]

for column in coverage_columns:
    print(f"\n{column.upper()}")
    print(df[column].value_counts())
```

```python
# Visualize healthcare coverage

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, column in zip(axes, coverage_columns):
    sns.countplot(
        x=column,
        data=df,
        ax=ax
    )

    ax.set_title(f'{column} Distribution')
    ax.set_xlabel(column)
    ax.set_ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
# Analyze chronic condition variables

chronic_columns = [
    'nchronic',
    'lchronic'
]

for column in chronic_columns:
    print(f"\n{column.upper()}")
    print(df[column].value_counts())
```

```python
# Visualize chronic conditions

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for ax, column in zip(axes, chronic_columns):
    sns.countplot(
        x=column,
        data=df,
        ax=ax
    )

    ax.set_title(f'{column} Distribution')
    ax.set_xlabel(column)
    ax.set_ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
# Average doctor visits by gender

average_visits_gender = (
    df.groupby('gender')['visits']
      .mean()
      .reset_index()
)

average_visits_gender['visits'] = (
    average_visits_gender['visits'].round(3)
)

display(average_visits_gender)

plt.figure(figsize=(7, 5))

sns.barplot(
    x='gender',
    y='visits',
    data=average_visits_gender
)

plt.title('Average Doctor Visits by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Doctor Visits')

plt.show()
```

```python
# Average doctor visits by illness level

average_visits_illness = (
    df.groupby('illness')['visits']
      .mean()
      .reset_index()
)

average_visits_illness['visits'] = (
    average_visits_illness['visits'].round(3)
)

display(average_visits_illness)

plt.figure(figsize=(10, 6))

sns.barplot(
    x='illness',
    y='visits',
    data=average_visits_illness
)

plt.title('Average Doctor Visits vs Number of Illnesses')
plt.xlabel('Number of Illnesses')
plt.ylabel('Average Doctor Visits')

plt.show()
```

```python
# Average doctor visits by health condition

average_visits_health = (
    df.groupby('health')['visits']
      .mean()
      .reset_index()
)

average_visits_health['visits'] = (
    average_visits_health['visits'].round(3)
)

display(average_visits_health)

plt.figure(figsize=(10, 6))

sns.barplot(
    x='health',
    y='visits',
    data=average_visits_health
)

plt.title('Average Doctor Visits by Health Condition')
plt.xlabel('Health Condition')
plt.ylabel('Average Doctor Visits')

plt.show()
```

```python
# Average doctor visits by reduced activity

average_visits_reduced = (
    df.groupby('reduced')['visits']
      .mean()
      .reset_index()
)

average_visits_reduced['visits'] = (
    average_visits_reduced['visits'].round(3)
)

display(average_visits_reduced)

plt.figure(figsize=(10, 6))

sns.barplot(
    x='reduced',
    y='visits',
    data=average_visits_reduced
)

plt.title('Average Doctor Visits by Reduced Activity')
plt.xlabel('Reduced Activity')
plt.ylabel('Average Doctor Visits')

plt.show()
```

```python
# Relationship between gender and doctor visits

plt.figure(figsize=(10, 6))

sns.countplot(
    x='gender',
    hue='visits',
    data=df
)

plt.title('Distribution of Doctor Visits by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Records')
plt.legend(title='Number of Visits')

plt.show()
```

```python
# Relationship between illness and doctor visits

plt.figure(figsize=(10, 6))

sns.boxplot(
    x='illness',
    y='visits',
    data=df
)

plt.title('Doctor Visits by Number of Illnesses')
plt.xlabel('Number of Illnesses')
plt.ylabel('Doctor Visits')

plt.show()
```

```python
# Relationship between income and illness

plt.figure(figsize=(10, 6))

sns.boxplot(
    x='illness',
    y='income',
    data=df
)

plt.title('Income Distribution by Number of Illnesses')
plt.xlabel('Number of Illnesses')
plt.ylabel('Income')

plt.show()
```

```python
# Doctor visits by chronic condition

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for ax, column in zip(axes, chronic_columns):
    sns.countplot(
        x=column,
        hue='visits',
        data=df,
        ax=ax
    )

    ax.set_title(f'Doctor Visits by {column}')
    ax.set_xlabel(column)
    ax.set_ylabel('Count')
    ax.legend(title='Visits')

plt.tight_layout()
plt.show()
```

```python
# Select numerical variables

numerical_columns = [
    'visits',
    'age',
    'income',
    'illness',
    'reduced',
    'health'
]

correlation_matrix = df[numerical_columns].corr()

display(correlation_matrix)
```

```python
# Correlation heatmap

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Healthcare Variables')

plt.show()
```

```python
# Identify relationships with doctor visits

visit_correlations = (
    correlation_matrix['visits']
    .drop('visits')
    .sort_values(ascending=False)
)

print("Correlation of variables with doctor visits:")

display(
    visit_correlations.to_frame(
        name='Correlation with Visits'
    )
)
```
